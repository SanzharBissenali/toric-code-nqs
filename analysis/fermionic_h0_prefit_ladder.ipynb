# Fermionic TC at $h=0$ — the sign-quality ladder (L=2 PBC)

Energy trajectories against the exact ground state $E_0=-4L^3=-32$, in order of
how the wavefunction's **sign structure** is supplied:

1. **cold start** — SR from random init: freezes at the positive-sector optimum
   $-22.52$ ($\delta=29.6\%$). SR cannot *create* sign structure.
2. **polish from hold-out pre-fit** (62/64 sign classes, joint fit): $-31.303$
   ($\delta=2.2\%$). SR cannot *repair* sign defects.
3. **polish from full-coverage pre-fit** (64/64, joint fit, no head): $-31.644$
   ($\delta=1.1\%$). Even perfect signs drift when smeared through the trunk.
4. **polish from joint-fit with phase head** (64/64): $-30.146$ ($\delta=5.8\%$).
   Joint fitting smears signs across trunk+head — *more* fragile.
5. **polish from the ANALYTIC phase head** (GF(2)-solved θ, phase-silent trunk):
   $\mathbf{-32.0000}$ ($\delta=1.3\times10^{-7}$). Signs stored explicitly in
   θ survive the amplitude drain untouched; SR polishes to the exact GS.

Curves: `results/fermionic_h0/*.curve.json`.

```python
import json
import numpy as np
import matplotlib.pyplot as plt

BASE = "../results/fermionic_h0"
RUNS = [
    ("cold start (random init)",
     f"{BASE}/gridinv_fermionic_L2_PBC_hx0.0_hz0.0_n2x4_nh4-8_inv8-8_k2.curve.json"),
    ("polish: hold-out pre-fit (62/64 signs)",
     f"{BASE}/gridinv_fermionic_L2_PBC_h0_n2x4_nh4-8_inv8-8_k2_polishHOLDOUT.curve.json"),
    ("polish: full pre-fit (64/64, joint, no head)",
     f"{BASE}/gridinv_fermionic_L2_PBC_h0_n2x4_nh4-8_inv8-8_k2_polishFULL.curve.json"),
    ("polish: joint fit with phase head",
     f"{BASE}/gridinv_fermionic_L2_PBC_h0_n2x4_nh4-8_inv8-8_k2_ph_polishFULL.curve.json"),
    ("polish: ANALYTIC phase head (GF(2)-solved)",
     f"{BASE}/gridinv_fermionic_L2_PBC_h0_n2x4_nh4-8_inv8-8_k2_ph_polishANA.curve.json"),
]
E0, TRAP = -32.0, -22.521
curves = []
for label, path in RUNS:
    c = json.load(open(path))["curve"]
    curves.append((label, np.asarray(c["energy"], float)))
    print(f"{label:48s} n={len(c['energy']):4d}  final E = {c['energy'][-1]:.6f}")
```

```python
fig, ax = plt.subplots(figsize=(8.5, 5.4))
colors = plt.cm.plasma([0.08, 0.32, 0.52, 0.70, 0.88])

for (label, E), c in zip(curves, colors):
    ax.plot(np.arange(len(E)), E, color=c, lw=1.7, label=label)

ax.set_ylim(-33.9, -7.5)
ax.axhline(E0, color="k", ls="--", lw=1.2)
ax.text(0.01, E0 - 0.35, r"exact $E_0 = -4L^3 = -32$", transform=ax.get_yaxis_transform(),
        ha="left", va="top", fontsize=10)
ax.axhline(TRAP, color="grey", ls=":", lw=1.0)
ax.text(0.115, TRAP + 0.55, "positive-sector optimum $-22.52$",
        transform=ax.get_yaxis_transform(), ha="left", va="bottom",
        fontsize=9, color="grey")

ax.set_xlabel("SR step (serial, resumes concatenated)")
ax.set_ylabel(r"$\langle H \rangle$")
ax.set_title("fermionic TC, $L=2$ PBC, $h=0$ — what sign quality buys", fontsize=11)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, fontsize=8.5, loc="upper left", bbox_to_anchor=(0.08, 1.00))

# inset: polished plateaus vs exact (the -30.15 joint-head plateau sits above range)
axi = ax.inset_axes([0.58, 0.50, 0.40, 0.30])
for (label, E), c in zip(curves[1:], colors[1:]):
    axi.plot(np.arange(len(E)), E, color=c, lw=1.5)
axi.axhline(E0, color="k", ls="--", lw=1.0)
axi.set_ylim(-32.12, -31.1)
axi.set_xlim(15, 200)
axi.tick_params(labelsize=8)
axi.set_title("polished plateaus vs exact (ANA lies ON the line)", fontsize=8, pad=3)
axi.spines[["top", "right"]].set_visible(False)

fig.tight_layout()
# fig.savefig("figs/fermionic_h0_prefit_ladder.png", dpi=400, bbox_inches="tight")
plt.show()
```

## L=3 — the ghost-sector trap and the flux-penalty head (2026-08-07 night)

Same recipe at L=3 (sampled-class GF(2) solve, certificate 10,000/10,000 on unseen
classes): the analytic-sign polish **plateaus at E = −89.97** with a *perfect*
on-orbit state (100% sign accuracy, uniform support amplitudes). MCMC census:
99.8% of the sampled weight sits in **ghost flux sectors** — the token-blind trunk
cannot separate the $2^{45}$ flux cosets, chains start in a random coset, and
cluster moves conserve it. Fix: the **flux-penalty head** (`--flux_penalty 6`) —
Re $\log\psi$ −= κ per violated closed-surface flux parity (45 analytic GF(2)
Gauss-law masks, zero parameters). Attached *training-free* to the plateaued
checkpoint it already gives ⟨H⟩ = −107.916(31); a 150-step SR polish closes to
**E = −107.99999592(6), δ = 3.8×10⁻⁸** in ~40 steps.

```python
L3_RUNS = [
    ("polish from analytic signs (no penalty)", "gridinv_fermionic_L3_PBC_h0_n2x4_nh4-8_inv8-8_k2_ph_polishANA"),
    ("+ flux penalty $\\kappa=6$ (warm from plateau)", "gridinv_fermionic_L3_PBC_h0_n2x4_nh4-8_inv8-8_k2_ph_fp6_polishANA"),
]
fig, ax = plt.subplots(figsize=(8.5, 5.0))
cl3 = plt.cm.plasma([0.25, 0.72])
off = 0
for (lbl, name), c in zip(L3_RUNS, cl3):
    d = json.load(open(f"{BASE}/{name}.curve.json"))["curve"]
    steps = np.asarray(d["step"]) + off
    ax.plot(steps, d["energy"], color=c, lw=1.6, label=lbl)
    off = steps[-1] + 1
ax.axhline(-108, ls="--", c="k", lw=1.0)
ax.text(6, -107.4, "exact $E_0=-108$", fontsize=9)
ax.axhline(-89.97, ls=":", c="0.45", lw=1.0)
ax.text(6, -89.2, "ghost-sector plateau (99.8% off-sector weight)", fontsize=9, color="0.35")
ax.annotate("training-free penalty attach:\n$\\langle H\\rangle = -107.916(31)$",
            xy=(302, -107.0), xytext=(180, -100.5), fontsize=9,
            arrowprops=dict(arrowstyle="->", lw=0.8))
ax.set_xlabel("SR step (runs concatenated)")
ax.set_ylabel(r"$\langle H \rangle$")
ax.set_title(r"$L=3$ fermionic TC, $h=0$: sign head alone vs sign head + flux penalty")
ax.legend(frameon=False, loc="center right", fontsize=9)
ax.spines[["top", "right"]].set_visible(False)
fig.tight_layout()
# fig.savefig("figs/fermionic_h0_L3_ghost.png", dpi=400, bbox_inches="tight")
d = json.load(open(f"{BASE}/{L3_RUNS[1][1]}.json"))["observables"]
print(f"final: E = {d['E0']:.8f} ± {d['E_err']:.1e}   delta = {d['delta']:.2e}   Vscore = {d['Vscore']:.2e}")
```
