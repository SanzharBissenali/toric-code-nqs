# Fermionic TC at $h=0$ — the sign-quality ladder (L=2 PBC)

Energy trajectories of the three SR runs against the exact ground state
$E_0=-4L^3=-32$:

1. **cold start** — SR from random init (the original benchmark run, including
   its guard-off continuation): freezes at the positive-sector optimum $-22.52$.
2. **polish from the hold-out pre-fit** (62/64 sign classes correct): freezes at
   $-31.303$ ($\delta=2.2\%$).
3. **polish from the full-coverage pre-fit** (64/64 correct at warm start):
   freezes at $-31.644$ ($\delta=1.1\%$) — the early off-support amplitude
   drain damages the phases slightly, and SR never repairs signs.

Curves: `results/fermionic_h0/*.curve.json` (re-run this notebook after
re-`rsync`ing to refresh a still-running curve).

```python
import json
import numpy as np
import matplotlib.pyplot as plt

BASE = "../results/fermionic_h0"
RUNS = [
    ("cold start (random init, + guard-off continuation)",
     f"{BASE}/gridinv_fermionic_L2_PBC_hx0.0_hz0.0_n2x4_nh4-8_inv8-8_k2.curve.json"),
    ("SR polish from hold-out pre-fit (62/64 sign classes)",
     f"{BASE}/gridinv_fermionic_L2_PBC_h0_n2x4_nh4-8_inv8-8_k2_polishHOLDOUT.curve.json"),
    ("SR polish from full-coverage pre-fit (64/64)",
     f"{BASE}/gridinv_fermionic_L2_PBC_h0_n2x4_nh4-8_inv8-8_k2_polishFULL.curve.json"),
]
E0, TRAP = -32.0, -22.521
curves = []
for label, path in RUNS:
    c = json.load(open(path))["curve"]
    curves.append((label, np.asarray(c["energy"], float)))
    print(f"{label:55s} n={len(c['energy']):4d}  final E = {c['energy'][-1]:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(8.5, 5.2))
colors = plt.cm.plasma([0.10, 0.50, 0.78])

for (label, E), c in zip(curves, colors):
    ax.plot(np.arange(len(E)), E, color=c, lw=1.7, label=label)

ax.set_ylim(-33.9, -7.5)
ax.axhline(E0, color="k", ls="--", lw=1.2)
ax.text(0.01, E0 - 0.35, r"exact $E_0 = -4L^3 = -32$", transform=ax.get_yaxis_transform(),
        ha="left", va="top", fontsize=10)
ax.axhline(TRAP, color="grey", ls=":", lw=1.0)
ax.text(0.115, TRAP + 0.55, "positive-sector optimum $-22.52$",
        transform=ax.get_yaxis_transform(), ha="left", va="bottom",
        fontsize=9, color="grey")

ax.set_xlabel("SR step (serial, resumes concatenated)")
ax.set_ylabel(r"$\langle H \rangle$")
ax.set_title("fermionic TC, $L=2$ PBC, $h=0$ — what sign quality buys", fontsize=11)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, fontsize=9, loc="upper left", bbox_to_anchor=(0.10, 1.00))

# inset (upper right, clear of curves and labels): polished plateaus vs exact
axi = ax.inset_axes([0.58, 0.56, 0.40, 0.33])
for (label, E), c in zip(curves[1:], colors[1:]):
    axi.plot(np.arange(len(E)), E, color=c, lw=1.5)
axi.axhline(E0, color="k", ls="--", lw=1.0)
axi.set_ylim(-32.1, -31.1)
axi.set_xlim(15, 200)
axi.tick_params(labelsize=8)
axi.set_title("polished plateaus vs exact", fontsize=8, pad=3)
axi.spines[["top", "right"]].set_visible(False)

fig.tight_layout()
fig.savefig("figs/fermionic_h0_prefit_ladder.png", dpi=400, bbox_inches="tight")
plt.show()
```
